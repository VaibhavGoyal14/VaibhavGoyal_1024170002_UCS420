```python
import pandas as pd
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
print("Dataset created successfully.")
print(df)
```

```python
print("\na) Shape:")
print(df.shape)
```

```python
print("\nb) DataFrame Information:")
df.info()
```

```python
print("\nc) Descriptive Statistics:")
print(df.describe())
```

```python
print("\nd) First 5 rows:")
print(df.head(5))
print("\nLast 3 rows:")
print(df.tail(3))
```

```python
average_salary = df["Salary"].mean()
print("\ne-i) Average Salary:", average_salary)
total_bonus = df["Bonus"].sum()
print("e-ii) Total Bonus:", total_bonus)
youngest_age = df["Age"].min()
print("e-iii) Youngest Employee's Age:", youngest_age)
highest_rating = df["Rating"].max()
print("e-iv) Highest Performance Rating:", highest_rating)
```

```python
df_sorted = df.sort_values(by="Salary", ascending=False)
print("\nf) DataFrame sorted by Salary (Descending):")
print(df_sorted)
```

```python
def rating_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
df["Performance_Category"] = df["Rating"].apply(rating_category)
print("\ng) Performance Category:")
print(df[["Name", "Rating", "Performance_Category"]])
```

```python
print("\nh) Missing Values:")
print(df.isnull().sum())
```

```python
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print("\ni) After renaming Employee_ID to ID:")
print(df)
```

```python
print("\nj-i) Employees with more than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])
print("\nj-ii) Employees in IT department:")
print(df[df["Department"] == "IT"])
```

```python
df["Tax"] = df["Salary"] * 0.10
print("\nk) DataFrame after adding Tax:")
print(df)
```

```python
df.to_csv("modified_employees.csv", index=False)
print("\nl) Modified DataFrame saved as modified_employees.csv")
```
