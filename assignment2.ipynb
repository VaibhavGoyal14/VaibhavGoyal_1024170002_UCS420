```python
L = [10, 0, 20, 40, 10, 70, 0, 0, 0, 20]
print("Original List L:", L)
L.append(50)
print("\nAfter append(50):", L)
L.insert(3, 60)
print("After insert(3, 60):", L)
L.remove(10)
print("\nAfter remove(10):", L)
L.pop(2)
print("After pop(2):", L)
L.sort()
print("\nAscending Order:", L)
L.sort(reverse=True)
print("Descending Order:", L)
print("\nFirst three elements:", L[:3])
print("Last three elements:", L[-3:])
average = sum(L) / len(L)
new_list = [x for x in L if x > average]
print("\nAverage =", average)
print("Elements greater than average:", new_list)
```

```python
scores = (10, 0, 20, 40, 10, 70, 0, 0)
print("Scores Tuple:", scores)
highest = max(scores)
lowest = min(scores)
print("\nHighest Score:", highest)
print("Index of Highest Score:", scores.index(highest))
print("Lowest Score:", lowest)
print("Lowest Score appears", scores.count(lowest), "times")
reversed_list = list(scores[::-1])
print("\nReversed List:", reversed_list)
x = int(input("\nEnter a score to search: "))
if x in scores:
    print("First occurrence index:", scores.index(x))
else:
    print("Score not present.")
try:
    scores[0] = 100
except TypeError as e:
    print("\nError:", e)
first, second, *remaining = scores
print("\nFirst Score:", first)
print("Second Score:", second)
print("Remaining Scores:", remaining)
```

```python
import random
random.seed(1024170002)
numbers = []
for i in range(100):
    numbers.append(random.randint(100, 900))
print("Random Numbers:")
print(numbers)
odd = []
for num in numbers:
    if num % 2 != 0:
        odd.append(num)
print("\nOdd Numbers:")
print(odd)
print("Count of Odd Numbers:", len(odd))
even = []
for num in numbers:
    if num % 2 == 0:
        even.append(num)
print("\nEven Numbers:")
print(even)
print("Count of Even Numbers:", len(even))
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_list = [num for num in numbers if is_prime(num)]
print("\nPrime Numbers:")
print(prime_list)
print("Count of Prime Numbers:", len(prime_list))
most_frequent = max(set(numbers), key=numbers.count)
print("\nMost Frequent Number:", most_frequent)
print("Frequency:", numbers.count(most_frequent))
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 0]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A =", A)
print("Set B =", B)
print("\nUnion of A and B:")
print(A.union(B))
print("\nIntersection of A and B:")
print(A.intersection(B))
print("\nA - B:")
print(A.difference(B))
print("\nB - A:")
print(B.difference(A))
# difference() returns elements present in one set but not the other,
# whereas symmetric_difference() returns elements present in either set but not both.
print("\nSymmetric Difference:")
print(A.symmetric_difference(B))
print("\nIs A subset of B?")
print(A.issubset(B))
print("\nIs B superset of A?")
print(B.issuperset(A))
x = int(input("\nEnter a value to remove from Set A: "))
A.discard(x)
print("Set A after discard():")
print(A)
# discard() does not raise an error if the element is not present,
# whereas remove() raises a KeyError.
```

```python
my_dict = {
    "name": "Vaibhav Goyal",
    "roll_no": "1024170002",
    "branch": "COPC",
    "age": 20,
    "city": "Patiala"
}
print("Original Dictionary:")
print(my_dict)
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming city to location:")
print(my_dict)
my_dict["cgpa"] = 9.62
print("\nAfter adding CGPA:")
print(my_dict)
my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)
dict1 = my_dict.copy()
removed = dict1.pop("branch")
print("\nAfter pop():")
print(dict1)
print("Removed Value:", removed)
dict2 = my_dict.copy()
del dict2["branch"]
print("\nAfter del:")
print(dict2)
# pop() returns the removed value, whereas del only deletes the key and returns nothing.
print("\nDictionary Items:")
for key, value in my_dict.items():
    print(key, "->", value)
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")
friend_dict = {
    "name": "Rahul",
    "roll_no": "1024170015",
    "branch": "COE",
    "age": 21,
    "location": "Delhi"
}
merged_dict = {**my_dict, **friend_dict}
print("\nMerged Dictionary:")
print(merged_dict)
# If both dictionaries have the same key, the value from the second dictionary (friend_dict) is kept.
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("\nDictionary with only string values:")
print(string_dict)
```
