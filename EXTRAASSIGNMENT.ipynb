```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("employee_dataset.csv")

print("Original Dataset:")
print(df)
```

```python
print("\nFirst 5 rows:")
print(df.head())

print("\nLast 5 rows:")
print(df.tail())
```

```python
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)
```

```python
print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("\nEmployee Names:")
print(df["Name"])

print("\nName and Salary:")
print(df[["Name", "Salary"]])

print("\nFirst 5 Rows using iloc:")
print(df.iloc[:5])

print("\nFirst 5 Rows, Name and Salary:")
print(df.loc[:4, ["Name", "Salary"]])
```

```python
print("\nEmployees with Salary > 60000:")
print(df[df["Salary"] > 60000])

print("\nEmployees with Experience >= 5:")
print(df[df["Experience"] >= 5])

print("\nIT Department Employees:")
print(df[df["Department"] == "IT"])

```

```python
print("\nEmployees Sorted by Salary:")
print(df.sort_values("Salary"))

print("\nEmployees Sorted by Salary Descending:")
print(df.sort_values("Salary", ascending=False))
```

```python
print("\nMissing Values:")
print(df.isnull().sum())


df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["PerformanceScore"] = df["PerformanceScore"].fillna(
    df["PerformanceScore"].median()
)

df["City"] = df["City"].fillna(df["City"].mode()[0])

print("\nMissing Values After Filling:")
print(df.isnull().sum())
```

```python
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)
```

```python
print("\nEmployees per Department:")
print(df["Department"].value_counts())

print("\nEmployees per City:")
print(df["City"].value_counts())
```

```python
print("\nAverage Salary by Department:")
print(df.groupby("Department")["Salary"].mean())

print("\nMaximum Salary by Department:")
print(df.groupby("Department")["Salary"].max())

print("\nAverage Performance by Department:")
print(df.groupby("Department")["PerformanceScore"].mean())

```

```python
df["Salary_per_Experience"] = df["Salary"] / (df["Experience"] + 1)

print("\nDataset with New Column:")
print(df)
```

```python
salary_array = np.array(df["Salary"])

print("\nNumPy Salary Array:")
print(salary_array)

print("\nTotal Salary:")
print(np.sum(salary_array))

print("\nAverage Salary:")
print(np.mean(salary_array))

print("\nMinimum Salary:")
print(np.min(salary_array))

print("\nMaximum Salary:")
print(np.max(salary_array))

print("\nStandard Deviation of Salary:")
print(np.std(salary_array))

print("\nSorted Salaries:")
print(np.sort(salary_array))
```

```python
age_array = np.array(df["Age"])

print("\nAge Array:")
print(age_array)

print("\nEmployees Above Age 25:")
print(age_array[age_array > 25])

print("\nTotal Age:")
print(np.sum(age_array))

print("\nAverage Age:")
print(np.mean(age_array))
```

```python
print("\nCorrelation Matrix:")
print(df[["Age", "Salary", "Experience", "PerformanceScore"]].corr())
```

```python
print("\nUnique Departments:")
print(df["Department"].unique())

print("\nNumber of Unique Departments:")
print(df["Department"].nunique())

print("\nUnique Cities:")
print(df["City"].unique())
```

```python
renamed_df = df.rename(columns={
    "PerformanceScore": "Performance",
    "Years_Experience": "Experience"
})

print("\nRenamed Columns:")
print(renamed_df.head())
```

```python
highest_paid = df.loc[df["Salary"].idxmax()]

print("\nHighest Paid Employee:")
print(highest_paid)
```

```python
df.to_csv("cleaned_employee_dataset.csv", index=False)

print("\nCleaned dataset saved successfully.")
```
